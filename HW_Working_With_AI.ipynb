## 2. AI is a tool, in a long line of tools
The calculator, the spreadsheet, statistics in a few clicks, statistical software, raster operations pixel by pixel, and now algorithms written as code. All of this is code that can be **read**. Python has taken a huge place in data processing over the last decades, and AI now speeds up writing it: autocompletion writes the syntax, and the AI proposes the structure (functions, their parameters, what they return).

So your job is not to type Python. It is to **read** it, understand what happens at its core, and decide. For any code an AI gives you, ask four questions:
1. Where in the code is the actual processing done (the "heart")?
2. Which library does it, and with which parameters?
3. What goes in, and what comes out?
4. How could I check that the result is right?

**Mini vocabulary** (run the cell below and read it line by line):
- a *function* is defined with **parameters** (`def f(x)`) and called with **arguments** (`f(3)`);
- it can *return* one value or several: several values come back as a **tuple**;
- a variable holds an **object** (a DataFrame, a GeoDataFrame, a raster...). An object has **attributes** (for example `gdf.crs`, the projection). Not every object carries its projection: a plain numpy array does not;
- a tuple cannot be modified after creation; to collect results and add to them, use a list or a dictionary.

```python
def describe(values):                    # "values" is a PARAMETER
    return min(values), max(values), sum(values) / len(values)   # returns ONE tuple with 3 values

low, high, mean = describe([3, 5, 10])   # [3, 5, 10] is the ARGUMENT; the tuple is unpacked into 3 variables
print(low, high, mean)

result = describe([3, 5, 10])
print(type(result), result)
```

## 3. How I work with an AI (and why it may work for you)
I think aloud, unstructured: with a coffee, in front of the TV, walking. I dictate, the AI transcribes and **structures** my scattered ideas, and understands what project it describes. For that the ideas have to be expressed, so say a lot. Then I ask questions, again and again, and once a result is proposed I question the result too.

Three habits that help:
- **Dictate, then reread.** Speech recognition garbles technical words (a satellite name can become something else). Reread before sending.
- **Make the AI ask first.** End with: *"Restate what you understood, then ask me your questions before proposing anything."*
- **Ask for alternatives, not one plan:** two or three approaches with pros and cons, and what the AI is unsure about.

## 4. What an AI does, and what it does not do
- **By default it writes from what it learned in training**, up to a cutoff date. It does not browse GitHub or the web unless it has tools for that. An agent *can* use tools (read a web page, search a named GitHub repository, run code) when they are enabled; see the documentation links in exercise E. It does not follow GitHub on its own.
- **It can be wrong in a convincing way**: invented datasets, references, function names, web links. It also tends to agree with you. A more powerful model, or a higher "effort" setting, usually reasons more carefully on multi-step tasks (an audit, for example), but it is not infallible.
- **A second AI reviewing the first is useful, but it is not independent proof.** Models can share blind spots, and a reviewer asked to "criticise" may invent problems. What really validates a result: run the code, compare with the source data, reproduce a known result, read the numbers yourself.

For this course, the quality of the science is secondary: what counts is that you learn the tools and keep control of what you produce.

**C.2 Fetch the data from the server (read and comment).**

```python
import os
import numpy as np
import matplotlib.pyplot as plt
import rasterio
from rasterio.warp import transform_bounds
from rasterio.windows import from_bounds
from pystac_client import Client

os.environ["GDAL_DISABLE_READDIR_ON_OPEN"] = "EMPTY_DIR"

bbox = (-0.46, 39.38, -0.30, 39.50)

catalog = Client.open("https://earth-search.aws.element84.com/v1")
search = catalog.search(collections=["sentinel-2-l2a"], bbox=bbox,
                        datetime="2024-10-30/2024-11-20",
                        query={"eo:cloud_cover": {"lt": 50}})
items = [i for i in search.items()
         if i.bbox[0] <= bbox[0] and i.bbox[1] <= bbox[1] and i.bbox[2] >= bbox[2] and i.bbox[3] >= bbox[3]]
item = min(items, key=lambda i: i.properties["eo:cloud_cover"])
print(item.id, item.datetime, item.properties["eo:cloud_cover"])


def read_band(name, shape=None):
    with rasterio.open(item.assets[name].href) as src:
        win = from_bounds(*transform_bounds("EPSG:4326", src.crs, *bbox), transform=src.transform)
        win = win.round_offsets().round_lengths()
        return src.read(1, window=win, out_shape=shape), src.window_transform(win), src.crs


red, transform, crs = read_band("red")
green, _, _ = read_band("green")
blue, _, _ = read_band("blue")
nir, _, _ = read_band("nir")
scl, _, _ = read_band("scl", shape=green.shape)

meta = item.assets["green"].extra_fields["raster:bands"][0]
red, green, blue, nir = [b * meta["scale"] + meta["offset"] for b in (red, green, blue, nir)]
print(green.shape, green.min(), green.max())
```

**C.3 Display and analyse (read and comment).**

```python
extent = (0, green.shape[1] * 0.01, green.shape[0] * 0.01, 0)

rgb = np.dstack([red, green, blue])
plt.imshow(np.clip(rgb / 0.3, 0, 1), extent=extent)
plt.title("True colour (red, green, blue)")
plt.xlabel("km")
plt.ylabel("km")
plt.show()

plt.imshow(green, cmap="gray", vmin=0, vmax=0.3, extent=extent)
plt.title("Green band (B03)")
plt.colorbar(label="Reflectance")
plt.show()

print("Mean green reflectance:", green.mean())
plt.hist(green.ravel(), bins=100, range=(0, 0.3))
plt.xlabel("Green reflectance")
plt.ylabel("Number of pixels")
plt.show()

bad = scl == 9
ndwi = (nir - green) / (nir + green)
ndwi[bad] = np.nan
water = ndwi > 0
print("Water fraction:", water[~bad].mean())

plt.imshow(ndwi, cmap="RdBu", vmin=-1, vmax=1, extent=extent)
plt.title("NDWI")
plt.colorbar(label="NDWI")
plt.show()
```
